```python
import requests

# Entrada de datos
nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()

# Llamado a la API
url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

# Verificación del estado de la respuesta
if respuesta.status_code == 200:
    datos = respuesta.json()

    # 1. Construcción de tipos sin comprensión de listas
    tipos = []
    for t in datos["types"]:
        tipos.append(t["type"]["name"])

    # 1. Construcción de habilidades sin comprensión de listas
    habilidades = []
    for h in datos["abilities"]:
        habilidades.append(h["ability"]["name"])

    # Característica adicional interesante: Extraer estadísticas base
    stats = {}
    total_stats = 0
    for s in datos["stats"]:
        nombre_stat = s["stat"]["name"]
        valor_stat = s["base_stat"]
        stats[nombre_stat] = valor_stat
        total_stats += valor_stat

    # Modelado del estado del Pokémon
    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos,
        "habilidades": habilidades,
        "stats": stats,
        "total_stats": total_stats
    }

    # Salida del estado actual
    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura'] / 10} m")
    print(f"Peso: {pokemon['peso'] / 10} kg")

    # 2. Condicional para clasificar por tipo
    if len(pokemon["tipos"]) > 1:
        print(f"Tipos: {', '.join(pokemon['tipos'])} (Pokémon de tipo mixto)")
    else:
        print(f"Tipo: {pokemon['tipos'][0]} (Pokémon de tipo único)")

    # 2. Condicional para clasificar por peso
    if pokemon["peso"] > 100:
        print("Clasificación de peso: Pokémon de gran tamaño")
    else:
        print("Clasificación de peso: Pokémon ligero")

    # 3. Listar las habilidades una por una usando un ciclo for
    print("\nHabilidades aprendidas:")
    for hab in pokemon["habilidades"]:
        print(f"  - {hab.capitalize()}")

    # 4. Uso adicional de la API: Evaluación del nivel de combate por estadísticas base
    print(f"\nEstadísticas base (Suma total: {pokemon['total_stats']}):")
    print(f"  - HP: {pokemon['stats']['hp']}")
    print(f"  - Ataque: {pokemon['stats']['attack']}")
    print(f"  - Defensa: {pokemon['stats']['defense']}")

    if pokemon["total_stats"] >= 500:
        print("Categoría de combate: Pokémon de alto rendimiento / Legendario o Etapa Final")
    elif pokemon["total_stats"] >= 300:
        print("Categoría de combate: Pokémon de nivel intermedio")
    else:
        print("Categoría de combate: Pokémon etapa inicial / Base")

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```
