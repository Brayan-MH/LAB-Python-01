```python
suma_pares = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i

print("Suma de los pares:", suma_pares)
```

```python
suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares = suma_pares + i
        print(f"Iteración i = {i:2d} | Es PAR   -> suma_pares = {suma_pares:2d}, suma_impares = {suma_impares:2d}")
    else:
        suma_impares = suma_impares + i
        print(f"Iteración i = {i:2d} | Es IMPAR -> suma_pares = {suma_pares:2d}, suma_impares = {suma_impares:2d}")

estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print("\n--- ESTADO FINAL ---")
print(estado_final)
```

El estado del programa está determinado por los valores de las variables (i, suma_pares, suma_impares).

Estado inicial: los acumuladores suma_pares y suma_impares comienzan en 0.
Durante la ejecución: en cada iteración, i toma valores del 1 al 10. La estructura if-else decide qué acumulador actualizar:
Si i es par, se suma a suma_pares.
Si i es impar, se suma a suma_impares.
Estado final: al terminar el bucle, los acumuladores contienen las sumas calculadas y se almacenan en un diccionario: {'suma_pares': 30, 'suma_impares': 25}.

El programa mantiene y actualiza dos acumuladores según la paridad de cada número, hasta obtener las sumas finales de pares e impares.
