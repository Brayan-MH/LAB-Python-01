```python
n = int(input("1. Ingrese el valor de n: "))
suma = 0

for i in range(1, n + 1):
    suma = suma + i
    print(f"Iteración {i}: suma acumulada = {suma}")

print(f"\nResultado final: La suma de los primeros {n} números naturales es {suma}")
```

```python
limite = int(input("2. Ingrese el límite hasta donde desea evaluar: "))

print(f"\nNúmeros pares entre 1 y {limite}:")
for i in range(1, limite + 1):
    if i % 2 == 0:
        print(f"El número {i} es par")
```

```python
limite = int(input("3.Ingrese el límite máximo: "))
suma_pares = 0

for i in range(1, limite + 1):
    if i % 2 == 0:
        suma_pares = suma_pares + i
        print(f"Número par encontrado: {i} | Suma acumulada = {suma_pares}")

print(f"\nResultado final: La suma de todos los números pares es {suma_pares}")
```

```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

print("Temperaturas:", temperaturas)

suma_total = 0.0
cantidad = 0

for temp in temperaturas:
    suma_total = suma_total + temp
    cantidad = cantidad + 1

promedio = suma_total / cantidad
print(f"Temperatura promedio: {promedio:.2f}°C")

print("\nTemperaturas mayores al promedio:")
for temp in temperaturas:
    if temp > promedio:
        print(f"- {temp}°C")
```
