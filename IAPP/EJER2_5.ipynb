```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)
```

```python
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break

    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    # 1.
    if precio < 0 or cantidad < 0:
        print(" Error: El precio y la cantidad deben ser mayores o iguales a 0. Entrada omitida.\n")
        continue


    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("\n--- RESUMEN DEL INVENTARIO ---")
# 3.
print("Estado final del inventario:", productos)
print(f"Total del inventario: ${total:.2f}")

# 2.
if len(productos) > 0:
    producto_mas_caro = max(productos, key=lambda p: p["precio"])
    print(f"Producto más caro: {producto_mas_caro['nombre']} (${producto_mas_caro['precio']:.2f})")
else:
    print("No se ingresaron productos válidos.")
```

El estado del programa está determinado por los valores de las variables (i, suma_pares, suma_impares).

Estado inicial: los acumuladores suma_pares y suma_impares comienzan en 0.
Durante la ejecución: en cada iteración, i toma valores del 1 al 10. La estructura if-else decide qué acumulador actualizar:
Si i es par, se suma a suma_pares.
Si i es impar, se suma a suma_impares.
Estado final: al terminar el bucle, los acumuladores contienen las sumas calculadas y se almacenan en un diccionario: {'suma_pares': 30, 'suma_impares': 25}.

El programa mantiene y actualiza dos acumuladores según la paridad de cada número, hasta obtener las sumas finales de pares e impares.
