```python
contador = 0
while contador < 5:
    print("Contador:", contador)
    contador += 1
```

```python
contador = 5
while contador > 0:
    print("1. Contador:", contador)
    contador -= 1
```

```python
numero = int(input("2. Ingrese un número (0 para salir): "))

while numero != 0:
    print(f"Ingresó el número: {numero}")
    numero = int(input("Ingrese otro número (0 para salir): "))

print("Programa finalizado por el usuario.")
```

El estado de un ciclo es el valor de su variable de control (por ejemplo, contador o numero) en cada iteración. Su funcionamiento se basa en tres elementos:

Inicialización: se establece el valor inicial de la variable antes de comenzar el ciclo.
Evaluación: la condición del while revisa el estado actual para decidir si continúa o finaliza la ejecución.
Actualización: la variable se modifica en cada iteración para acercarse a la condición de salida y evitar bucles infinitos.

De esa manera un ciclo funciona porque inicializa un estado, lo evalúa repetidamente y lo actualiza hasta que se cumple la condición de terminado.
