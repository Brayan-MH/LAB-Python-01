```python
# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

# Operaciones básicas
print("Temperaturas:", temperaturas)
```

```python
temperaturas.pop(0)
print("2.", temperaturas)
```

```python
temperaturas.insert(1,23)
print("3.",temperaturas)
```

```python
temperaturas_ord = list(temperaturas)
temperaturas_ord.sort ()
print("4.", temperaturas_ord)
```

```python
suma_temp = 0
temp_max = temperaturas[0]
temp_min = temperaturas[0]
cant = 0
for n in temperaturas:
  suma_temp += n
  if n>temp_max:
    temp_max = n
  if n<temp_min:
    temp_min = n
  cant += 1

diferencia = temp_max - temp_min
prom = suma_temp/cant

print(f"5.Temperatura mas alta: {temp_max}")
print(f"  Temperatura mas baja: {temp_min}")
print(f"  Diferencia de temperaturas: {diferencia}")
print(f"  Promedio de las temperaturas: {prom}")
```

```python
find = -1
actual = 0

while actual <len(temperaturas):
  if temperaturas[actual] == 23.7:
    find = actual
    break
  actual += 1

if find != 1:
  print(f"6. La temperatura buscada esta en el indice:{actual -1 } (Dentro de la posicion de la lista).")
else:
  print("La temperatura no esta presente en la lista.")
```
