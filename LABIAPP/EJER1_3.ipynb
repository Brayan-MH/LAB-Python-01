```python
# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]
```

```python
print("1.")
print(inventario[0]["nombre"])
```

```python
nue_producto = {"nombre":"Sal","precio":2000,"stock": 20}
inventario.append(nue_producto)

print("2.")

for i in inventario:
  print(i)
print()
```

```python
print("3.")
indice = 0
while indice < len(inventario):
  if inventario[indice]["nombre"] == "Leche":
    inventario [indice]["stock"] = inventario[indice]["stock"] - 2
    print(f"Inventario actualizado de Leche:{inventario[indice]["stock"]}")
    break
  indice += 1
print ()
```

```python
valor = 0
n = 0
for n in inventario:
  subtotal = n["precio"]*n["stock"]
  valor += subtotal
print("4. El valor total del inventario es:",valor)

```

```python
print("6 y 7.")
incremento = 1.10 #10% incremento
for n in inventario:
  nue_precio = n["precio"]*incremento
  red_nue_precio = round(nue_precio,2)
  print(f"{n['nombre']}:{n['precio']}","->",red_nue_precio)


```

```python
find = input("8.Ingrese el nombre del producto a buscar: ")

encontrado = False

for i in inventario:
    if i["nombre"].lower() == find.lower():
        print(f"Se encontró el producto.\n DETALLE\nPrecio: {i['precio']}\nStock: {i['stock']}")
        encontrado = True
        break

if encontrado == False:
    print("No se encontró el producto en el inventario.")
```

```python
inventario[2]["stock"] = 0

indice = 0
while indice < len(inventario):
    if inventario[indice]["stock"] == 0:
        producto_eliminado = inventario.pop(indice)
        print(f"Producto eliminado por stock agotado: {producto_eliminado['nombre']}")
    else:
        indice = indice + 1
print("9. Inventario tras revisar y eliminar agotados:")
for item in inventario:
    print(item)
print()

```

```python
n = len(inventario)
i = 0
while i < n - 1:
    j = 0
    while j < n - i - 1:
        if inventario[j]["precio"] > inventario[j + 1]["precio"]:
            temporal = inventario[j]
            inventario[j] = inventario[j + 1]
            inventario[j + 1] = temporal
        j = j + 1
    i = i + 1
print("10. Inventario ordenado por precio de menor a mayor:")
for producto in inventario:
    print(f" - {producto['nombre']}: ${producto['precio']} (Stock: {producto['stock']})")
```
