```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}
```

```python
print("1.")

contador_prestados = 0
contador_disponibles = 0

for libro in biblioteca["libros"]:
    if libro["prestado"] == True:
        contador_prestados = contador_prestados + 1
    else:
        contador_disponibles = contador_disponibles + 1

print(f"Libros prestados: {contador_prestados}")
print(f"Libros disponibles: {contador_disponibles}")
```

```python
print("2.")
nuevo_libro = {"titulo": "Cien años de soledad", "autor": "Gabriel García Márquez", "prestado": False}
biblioteca["libros"].append(nuevo_libro)
```

```python
print("3.")
indice = 0
while indice < len(biblioteca["libros"]):
    if biblioteca["libros"][indice]["titulo"] == "1984":
        biblioteca["libros"][indice]["prestado"] = True
        break
    indice = indice + 1
```

```python
print("4.")
print("\nEstado actualizado del sistema:")
print(biblioteca)
```
