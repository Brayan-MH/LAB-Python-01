TypeError: 'tuple' object does not support item assignment
¿Por qué ocurre?: Las tuplas en Python son estructuras de datos INMUTABLES.Esto significa que una vez creadas, sus elementos no pueden ser modificados, añadidos ni eliminados en su lugar.


```python
posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0.0
indice = 0

```

```python
print("4.")
while indice < len(posiciones) - 1:
    x1 = posiciones[indice][0]
    y1 = posiciones[indice][1]

    x2 = posiciones[indice + 1][0]
    y2 = posiciones[indice + 1][1]

    distancia_segmento = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
    distancia_total = distancia_total + distancia_segmento

    indice = indice + 1

print(f"Distancia total recorrida: {distancia_total:.2f}")
```
