```python
# Estado inicial
saldo = 500_000      # Saldo de la cuenta
deposito = 150_000   # Monto a depositar
retiro = 80_000      # Monto a retirar

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero

# Salida del estado final
print("Saldo final de la cuenta:", saldo)
```

```python
interes = 0.02 #2% mensual
```

```python
saldo_final = saldo + (saldo*interes)
```

```python
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

¿Qué representa el estado inicial y el estado final del programa?

El estado inicial representa los valores sin cambios de las variables antes de ejecutar las instrucciones de trabajo. Una vez ejecutadas las ordenes, el estado final cambia correspondiente a las modificaciones realizadas en el codigo.

Explica cómo cada instrucción afecta la memoria.


El estado inicial almacena los datos recolcetados en las variables. El interes crea un espacio de memoria donde se almacena para poder y usar para posteriosmente leer los datos almacenados y procesarlos para guardarlos en saldo y saldo_final realizando las operaciones que son requeridas y finalmente esos resultados guardados en la RAM se imprimen  en print que accede a esa memoria y la muestra al usuario.
